```python
#printing iris dataset
import numpy as np
import pandas as pd
import plotly.express as px
pd.set_option('display.max_columns',None)
pd.set_option('display.max_rows',None)
pd.set_option('display.width',None)
df=pd.DataFrame(px.data.iris())
print(df)
```

```python
#line plot
df_sorted=df.sort_values(by="sepal_length")
fig=px.line(df_sorted,x="sepal_length",y="sepal_width",color="species",title="Line plot of Sepal width vs Sepal length")
fig.show()
```

```python
#scatter plot
fig=px.scatter(df,x="sepal_length",y="sepal_width",color="species",title="Scatter plot of Sepal length vs Sepal width")
fig.show()
```

```python
#Area Plot
fig=px.area(df_sorted,x="sepal_length",y="sepal_width",color="species",title="Area plot of Sepal Width vs Sepal length")
fig.show()
```

```python
#Box Plot
fig=px.box(df,x="species",y="petal_length",color="species",title="Box plot of petal length by species")
fig.show()
```

**VIOLIN PLOT**- Similar to box plot,except that they also show the probability density of the data at different values

```python
#Violin plot
fig=px.violin(df,x="species",y="petal_length",color="species",box=True,points="all",title="Violin plot of petal length by species")
fig.show()
```

```python
#Histogram
fig=px.histogram(df,x="sepal_length",color="species",nbins=20,title="Histogram of Sepal length by species")
fig.show()
```

```python
#Bar plot
mean_petal_length=df.groupby("species")["petal_length"].mean().reset_index()
fig=px.bar(mean_petal_length,x="species",y="petal_length",color="species",title="Bar plot of Average Petal length by species")
fig.show()
```

```python
#pie plot
species_counts=df["species"].value_counts().reset_index()
species_counts.columns=["species","counts"]
fig=px.pie(species_counts,values="counts",names="species",title="Pie chart of species Distribution")
fig.show()
```

**SUNBURST PLOT**- It represents the hierarchial data as sectors laid out over several levels of concentric Rings.

```python
#Sunburst plot
fig=px.sunburst(df,path=["species"],values=None,title="Sunburst chart of species Distribution")
fig.show()
```

**TREE MAP**- It represents hierarchial data as nested rectangular sectors.

```python
import plotly.express as px
#Tree Map
fig=px.treemap(df,path=["species"],values=None,title="Treemap of species Distribution")
fig.show()
```

```python
#Parallel coordinates
df["species_id"]=df["species"].astype('category').cat.codes
fig=px.parallel_coordinates(df,color="species_id",dimensions=["sepal_length","sepal_width","petal_length","petal_width"],color_continuous_scale=px.colors.diverging.Tealrose)
fig.show()
```

```python
#Density Heat Map
fig=px.density_heatmap(df,x="sepal_length",y="sepal_width",marginal_x="histogram",marginal_y="histogram",title="Density Heat Map of speaf Dimensions")
fig.show()
```

**CONTOUR PLOT**- Used to represent 3D data in 2D using contour lines

```python
import plotly.express as px
import seaborn as sns
import pandas as pd
df=pd.DataFrame(px.data.iris())
fig=px.density_contour(df,x="sepal_length",y="sepal_width",color="species",title="Contour plot of Sepal length vs Sepal width by species")
fig.update_traces(contours_coloring="fill",showscale=True)
fig.show()
```

```python
#Scatter points to the contour plot
fig=px.density_contour(df,x="sepal_length",y="sepal_width",color="species")
fig.update_traces(contours_coloring="fill",showscale=True)
fig.update_layout(title="Contour Plot with Data Points")
fig.add_scatter(x=df["sepal_length"],y=df["sepal_width"],mode='markers',marker=dict(color='black',size=3),name='Data Points')
fig.show()
```

**3D PLOTS**

```python
#3D Scatter Plot
import plotly.express as px
import seaborn as sns
import pandas as pd
df=sns.load_dataset("iris")
fig=px.scatter_3d(df,x="sepal_length",y="sepal_width",z="petal_length",color="species",title="3D Scatter plot: Sepal length vs sepal width vs petal length")
fig.show()
```

```python
#3D Line plot
df_sorted=df.sort_values(by="sepal_length")
fig=px.line_3d(df_sorted,x="sepal_length",y="sepal_width",z="petal_length",color="species",title="3D Line plot:Sepal & Petal relationships")
fig.show()
```

```python
#3D Mesh Plot
import plotly.graph_objects as go
import pandas as pd
import numpy as np
df_setosa=df[df["species"]=="setosa"]
fig=go.Figure(
    data=[go.Mesh3d(x=df_setosa["sepal_length"],y=df_setosa["sepal_width"],z=df_setosa["petal_length"],
                    opacity=0.5,
                    color='Light blue',
                    name='setosa 3D Mesh'
                    )]
)
fig.update_layout(title='3D MeshPlot: Setosa sepal vs Petal Dimensions')
fig.show()
```

```python
#Surface Plot
from scipy.interpolate import griddata
import numpy as np
import plotly.graph_objects as go # Added missing import

#Grid Setup
x=df["sepal_length"]
y=df["sepal_width"]
z=df["petal_length"]
xi=np.linspace(x.min(),x.max(),50)
yi=np.linspace(y.min(),y.max(),50)
xi,yi=np.meshgrid(xi,yi)
zi=griddata((x,y),z,(xi,yi),method='cubic') # Corrected this line

#Create Surface plot
fig=go.Figure(data=[go.Surface(x=xi,y=yi,z=zi,colorscale='Viridis')]) # Changed go.surface to go.Surface
fig.update_layout(title='3D Surface Plot: Sepal to petal length Interpolation') # Corrected typo in title
fig.show()
```

```python
 #Bar chart- All Average Measurements
avg_df=(df.groupby("species")[["sepal_length","sepal_width","petal_length","petal_width"]].mean().reset_index())
print(avg_df)
melted=avg_df.melt(id_vars="species",var_name="measurement",value_name="average")
fig=px.bar(melted,x="species",y="average",color="measurement",barmode="group",title="Average iris measurements by species")
fig.show()
```
